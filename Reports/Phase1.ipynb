# 1. Project Goal
The goal of this project is to apply data mining techniques to the Diabetes Health Dataset. The project aims to build a classification model to predict individuals diabetes status based on the available attributes in the dataset. It also aims to cluster individuals into groups based on similarities in their available attributes. The results of the classification and clustering tasks will be evaluated and analyzed to identify meaningful patterns in the dataset.
